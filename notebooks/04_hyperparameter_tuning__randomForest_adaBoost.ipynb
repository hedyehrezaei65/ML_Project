# Hyperparameter Tuning

```python
#hyper parameter tuning
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import RandomizedSearchCV

import pandas as pd 
import numpy as np
```

```python
train_X_final = pd.read_csv("../data/train_test_data/train_X_final.csv")
test_X_final = pd.read_csv("../data/train_test_data/test_X_final.csv")
y_train_encoded = pd.read_csv("../data/train_test_data/y_train_encoded.csv")


train_X_final.head()
```

```python
RandomForestClassifier().get_params()
```

# Random Forest

```python

rf = RandomForestClassifier()


param_dist_0 = {
    "n_estimators": [ 100,200],
    "max_depth": [35 , 45 ],          
    "min_samples_leaf": [ 20 ,50],  
    "max_features": ["sqrt", "log2"],
}

param_dist_1 = {
    "n_estimators": [100, 200, 500],
    "max_depth": [20 , 30 , 40],
    "max_leaf_nodes": [ 5000, 10000 ,20000],
    "max_features": ["sqrt", "log2"]

}

param_dist_2 = {
    'n_estimators': [100, 200,300],        # 200, 300, 400, 500, 600
    'max_depth': [ 30, 35,40],               # Removed None and 70+
    'max_features': ['sqrt', 'log2', 0.6],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True]                              # Often faster + good enough
}

search = RandomizedSearchCV(
    rf,
    param_distributions=param_dist_2,
    n_iter=30,
    cv=3,
    scoring="f1_macro", #f1_macro
)

search.fit(train_X_final, y_train_encoded)

print(search.best_params_)
```

```python

adaBoost = AdaBoostClassifier()

param_dist = {
    "n_estimators": [100, 200, 500],
    "learning_rate": [20 , 30 , 40]
}

parameters = {'n_estimators':[50,100, 120],                
              'learning_rate':[0.1, 0.5, 1.0]               
              #'estimator__min_samples_split' : np.arange(2, 8, 2),               
              #'estimator__max_depth' : np.arange(1, 4, 1)              
             } 

search = RandomizedSearchCV(
    adaBoost,
    param_distributions=parameters,
    scoring="accuracy", #f1_macro
)

search.fit(train_X_final, y_train_encoded)

print(search.best_params_)
```
